```python
def harf_notu(puan:int):
    try:
        if (puan < 0 or puan > 100):
            raise ValueError ("Lütfen Puan olarak 0-100 arası bir tam sayı giriniz.")
        else:
            if puan >=90 and puan <=100:
                print("Harf Notunuz: AA")
            elif puan >=80 and puan <=89:
                print("Harf Notunuz: BA")
            elif puan >=70 and puan <=79:
                print("Harf Notunuz: BB")
            elif puan >=60 and puan <=69:
                print("Harf Notunuz: CB")
            elif puan >=50 and puan <=59:
                print("Harf Notunuz: CC")
            else:
                print("Harf Notunuz: FF, Dersi Geçemediniz.")
    except Exception as e:
        print(f"Hata: {e}")
    finally:
        print("İşlem Tamamlandı.")
try:
    puan = int(input("Lütfen Notunuzu Giriniz: "))
    harf_notu(puan)
except ValueError:
    print("Lütfen Sayısal Bir Değer Giriniz.")         
```
