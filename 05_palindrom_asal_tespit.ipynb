```python
def palindrom_kontrol(a:str) ->str:
    if a == a[::-1]:
        return f"{a} kelimesi bir palindromdür."
    else:
        return f"{a} kelimesi bir palindrom kelime değildir."
kelime = input("Lütfen Palindrom Kontrolü İçin Kelimenizi Giriniz: ").strip().lower()
sonuc = palindrom_kontrol(kelime)
print(sonuc)
```

```python
def asal_tespit(sayi):
    asal_mi = True
    if sayi <= 1:
        return False
    else:
        for i in range(2,sayi):
            if sayi % i == 0:
                asal_mi = False
                break
    return asal_mi
sonuc = asal_tespit(11)
print(sonuc)
```
