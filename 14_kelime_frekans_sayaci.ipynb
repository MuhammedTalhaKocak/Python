```python
from collections import Counter
from string import punctuation
```

```python
noktalama_isaretleri = punctuation
cumle = input("Cümleniz: ").lower()
saf_cumle =  "".join(char for char in cumle if char not in noktalama_isaretleri)
kelimeler = saf_cumle.split(" ")
counter = Counter(kelimeler)
print(f"En çok geçen 5 kelime: {counter.most_common(5)}")
```
