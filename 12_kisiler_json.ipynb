```python
import json
import re

try:
    with open("kisiler.json", "r", encoding="utf-8") as f:
        kisiler = json.load(f)
except FileNotFoundError:
    kisiler = []
```

```python
def kisi_ekle(kisiler):
    isim = input('Lütfen isminizi giriniz: ')
    telefon = input('Lütfen Telefon numaranızı giriniz: ')
    email = input('Lütfen mail adresinizi giriniz: ')
    pattern = r"^\w+@\w+\.com$"
    while True:
        if re.search(pattern, email):
            print('Geçerli bir mail adresi girdiniz.')
            break
        else:
            print('Lütfen geçerli bir mail adresi giriniz.')
            email = input('Lütfen mail adresinizi giriniz: ')
    kisi = {
        'İsim': isim,
        'Telefon': telefon,
        'Email': email
    }
    kisiler.append(kisi)
    print('Kişi Başarıyla Eklendi.')
    return kisiler
```

```python
def kisi_ara(kisiler):
    aranan_isim = input('Aramak istediğiniz ismi giriniz: ')
    sonuclar = [kisi for kisi in kisiler if kisi['İsim'] == aranan_isim]
    if sonuclar:
        for kisi in sonuclar:
            print(kisi)
    else:
        print('Bu isimde bir kişi bulunamadı.')
```

```python
def kisi_sil(kisiler):
    silinecek_isim = input('Silmek istediğiniz ismi giriniz: ')
    kisi_isimleri = [kisi['İsim'] for kisi in kisiler]
    if silinecek_isim not in kisi_isimleri:
        print('Böyle bir kişi bulunamadı.')
    else:
        kisiler = [kisi for kisi in kisiler if kisi['İsim'] != silinecek_isim]
    return kisiler
```

```python
def kisi_listele(kisiler):
    if not kisiler:
        print("Hiç kayıtlı kişi yok.")
    else:
        for kisi in kisiler:
            print(f"İsim: {kisi['İsim']}, Telefon: {kisi['Telefon']}, Email: {kisi['Email']}")
            print()
```

```python
while True:
    print(""" Telefon Rehberi Uygulamasına Hoşgeldiniz ...
    1 - Kişi Ekleme
    2 - Kişi Arama
    3 - Kişi Silme
    4 - Kişileri Listeleme
    5 - Çıkış
""")
    user = input('Seçiminiz: ')
    if user not in ["1", "2", "3", "4", "5"]:
        print('Lütfen geçerli bir seçim yapınız.')
    elif user == "1":
        kisiler = kisi_ekle(kisiler)
    elif user == "2":
        kisi_ara(kisiler)
    elif user == "3":
        kisiler = kisi_sil(kisiler)
    elif user == "4":
        kisi_listele(kisiler)
    elif user == "5":
        with open("kisiler.json", "w", encoding="utf-8") as f:
            json.dump(kisiler, f, ensure_ascii=False, indent=2)
        break
```
