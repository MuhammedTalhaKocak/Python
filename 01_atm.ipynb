```python
import time
def para_yatir(bakiye: int, miktar: int) -> int:
    bakiye = bakiye + miktar
    return bakiye
def para_cek(bakiye: int, miktar: int):
    if miktar < 0:
        return "Negatif tutarda para çekemezsiniz."
    elif miktar > bakiye:
        return None
    else:
        print("Para Çekme İşlemi yapılıyor ....")
        time.sleep(3)
        bakiye = bakiye - miktar
        return bakiye
bakiye = int(input("Lütfen Bakiyenizi Giriniz: "))
while True:
    user = input("Hesabınıza Giriş Yapıldı, devam etmek için evet devam etmek istemiyorsanız hayır yazınız: ")
    if user.lower() == 'hayır':
        print("Hoşçakalın ... Kartınızı Almayı unutmayınız.")
        break
    elif user.lower() == 'evet':
       islem = input("Para yatırmak için 'yatir', para çekmek için 'cek' yaziniz.") 
       if islem.lower() == 'cek':
            miktar = int(input("Lütfen çekmek istediğiniz para miktarını giriniz: "))
            sonuc = para_cek(bakiye, miktar)
            if isinstance(sonuc, str):
                print(sonuc)
            elif sonuc is None:
                print("Bakiyenizden fazla çekemezsiniz.")
            else:
                bakiye = sonuc
                print(f"Yeni Bakiyeniz: {bakiye}")
       elif islem.lower() == 'yatir':
            miktar = int(input("Lütfen yatırmak istediğiniz para miktarını giriniz: "))
            bakiye = para_yatir(bakiye, miktar)
            print(bakiye)
       else:
            print("Geçersiz İşlem.")
```
