```python
import time
def celsius_to_fahrenheit(c): 
    f = float(c) * (9/5) + (32)
    return f"{f:.2f}"
def fahrenheit_to_celsius(f): 
    c = (f - 32) * 5/9
    return f"{c:.2f}"
def celsius_to_kelvin(c): 
    k = c + 273.15
    return f"{k:.2f}"
def kelvin_to_celsius(k):
    c = k - 273.15
    return f"{c:.2f}"
print("*****Sıcaklık Dönüştürücü Programına Hoşgeldiniz.*****")
while True:
        islem_secimi = input("""
        "1: Celsius→Fahrenheit, 2: Fahrenheit→Celsius, 3: Celsius→Kelvin, 4: Kelvin→Celsius"
        """)
        try:
            if islem_secimi not in ["1", "2", "3", "4"]:
                print("Geçersiz bir seçim yaptınız.")
            elif islem_secimi == "1":
                derece = int(input("Celsius - Fahrenheit dönüşümü için değer giriniz: "))
                print("Sonuçlar Hesaplanıyor ...")
                time.sleep(2)
                sonuc = celsius_to_fahrenheit(derece)
                print(f"{derece} Celsius = {sonuc} Fahrenheit")
            elif islem_secimi == "2":
                derece = int(input("Fahrenheit - Celsius dönüşümü için değer giriniz: "))
                print("Sonuçlar Hesaplanıyor ...")
                time.sleep(2)
                sonuc = fahrenheit_to_celsius(derece)
                print(f"{derece} Fahrenheit = {sonuc} Celsius")
            elif islem_secimi == "3":
                derece = int(input("Celsius - Kelvin dönüşümü için değer giriniz: "))
                print("Sonuçlar Hesaplanıyor ...")
                time.sleep(2)
                sonuc = celsius_to_kelvin(derece)
                print(f"{derece} Celsius = {sonuc} Kelvin")
            elif islem_secimi == "4":
                derece = int(input("Kelvin - Celsius dönüşümü için değer giriniz: "))
                print("Sonuçlar Hesaplanıyor ...")
                time.sleep(2)
                sonuc = kelvin_to_celsius(derece)
                print(f"{derece} Kelvin = {sonuc} Celsius")
        except ValueError as e:
            print(f"Hata: {e}")
        finally:
            print("İşlem Tamamlandı.")
        durum = input("Devam etmek için evet, devam etmek istemiyorsanız hayır yazınız.").lower()
        if durum == "hayır":
            print("Program Sonlanıyor ...")
            time.sleep(3)
            print("Program Sonlanmıştır.Kendinize iyi bakın.")
            break
```
