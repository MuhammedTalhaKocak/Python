```python
notlar = []
puan_adedi = int(input("Kaç tane not girmek istiyorsunuz: "))
for i in range(0,puan_adedi):
    puan = int(input("Notunuz: "))
    notlar.append(puan)
ortalama = sum(notlar) / len(notlar)
print(f"Not Ortalamanız: {ortalama:.2f}")
if ortalama > 50:
    print("Geçtiniz")
else:
    print("Kaldınız")
```
