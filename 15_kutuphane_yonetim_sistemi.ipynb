```python
class Kitap:
    def __init__(self, ad, yazar):
        self.ad = ad
        self.yazar = yazar
        self.mevcut_mu = True
    def __str__(self):
        return f"{self.ad} ({self.yazar})"
    def __repr__(self):
        return self.__str__()
    def odunc_al(self):
        if self.mevcut_mu:
            self.mevcut_mu = False
            print(f"{self.ad} kitabını ödünç aldınız.")
        else:
            print(f"{self.ad} kitabı şu anda mevcut değil.")
    def iade_et(self):
        if not self.mevcut_mu: # kitap mevcut değilse iade edilebilir
            self.mevcut_mu = True # iade işlemi oluyor, kitap mevcut hale geliyor
            print(f"{self.ad} kitabını iade ettiniz.")
        else:
            print(f"{self.ad} kitabı zaten mevcut. İade edemezsiniz.")
```

```python
kitap1 = Kitap("Suç ve Ceza", "Fyodor Dostoyevski")
kitap2 = Kitap("1984", "George Orwell")
print(f'1.Kitap İsmi: {kitap1.ad}, Yazar: {kitap1.yazar}, Mevcut mu: {kitap1.mevcut_mu}')
print(f'2.Kitap İsmi: {kitap2.ad}, Yazar: {kitap2.yazar}, Mevcut mu: {kitap2.mevcut_mu}')
```

```python
kitap1.odunc_al() # Suç ve Ceza kitabını ödünç alıyoruz
kitap1.iade_et() # Suç ve Ceza kitabını iade ediyoruz
kitap1.iade_et() # Suç ve Ceza kitabını tekrar iade etmeye çalışıyoruz
```

```python
class Uye: # uye class'ı oluşturuluyor.
    def __init__(self, isim):
        self.isim = isim 
        self.mevcut_kitaplar = [] # üyenin mevcut kitapları listesi başlangıçta boş
    def odunc_kitap_al(self, kitap:Kitap): # üye ödünç kitap alacak.
        if kitap.mevcut_mu == True: # eğer kitap mevcutsa
            kitap.odunc_al() # kitabı ödünç al
            self.mevcut_kitaplar.append(kitap) # üyenin mevcut kitaplar listesine ekle
```

```python
uye1 = Uye("Talha")
uye1.odunc_kitap_al(kitap2)
print(uye1.mevcut_kitaplar)
```
