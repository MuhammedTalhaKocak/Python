```python
import random
rastgele_sayi = random.randint(0,100)
deneme_sayisi = 0
max_deneme = 10
while True:
    tahmin = int(input("Lütfen Tahmin Sayınızı Giriniz: "))
    deneme_sayisi += 1
    if tahmin == rastgele_sayi:
        print(f"Tebrikler, sayıyı {deneme_sayisi} denemede doğru tahmin ettiniz.")
        break
    elif deneme_sayisi == max_deneme:
        print("Deneme Hakkınız Kalmadı. Görüşmek Üzere.")
        break
    elif tahmin < rastgele_sayi:
        print("Lütfen daha yüksek bir sayı tahmin ediniz.")
    else:
        print("Lütfen daha küçük bir sayı tahmin ediniz.")
```
