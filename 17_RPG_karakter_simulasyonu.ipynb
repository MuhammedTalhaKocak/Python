```python
class Karakter:
    def __init__(self, isim, saldiri_gucu, can = 100):
        self.isim = isim
        self.saldiri_gucu = saldiri_gucu
        self.can = can
    def saldir(self ,hedef:Karakter):
        print(f'{self.isim} {hedef.isim} karakterine {self.saldiri_gucu} kadar hasar verdi.')
        hedef.can = hedef.can - self.saldiri_gucu 
        if hedef.can < 0: 
            print(f'{hedef.isim} karakteri öldü.')
class Buyucu(Karakter):
    def __init__(self, isim, saldiri_gucu, mana, can = 100):
        super().__init__(isim, saldiri_gucu, can)
        self.mana = mana
    def buyu_yap(self, hedef:Karakter):
        if self.mana >= 10:
            print(f'{self.isim} {hedef.isim} karakterine {self.saldiri_gucu * 2} kadar hasar verdi.')
            hedef.can = (hedef.can) - (self.saldiri_gucu * 2)
            self.mana = self.mana - 10
        else:
            print(f'{self.isim} {hedef.isim} karakterine {self.saldiri_gucu} kadar hasar verdi.')
            hedef.can = (hedef.can) - (self.saldiri_gucu)

buyucu1 = Buyucu("Merlin", 15, 25)
dusman1 = Karakter("Ork", 10)
buyucu1.buyu_yap(dusman1)
print(f"{dusman1.isim} canı: {dusman1.can}")
buyucu1.buyu_yap(dusman1)
print(f"{dusman1.isim} canı: {dusman1.can}")
buyucu1.buyu_yap(dusman1)
print(f"{dusman1.isim} canı: {dusman1.can}")
```
