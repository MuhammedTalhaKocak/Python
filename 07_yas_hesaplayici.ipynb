```python
from datetime import datetime
try:
    dogum_yili = int(input("Lütfen Doğum Yılınızı Giriniz: "))
    simdi = datetime.now()
    bu_yil = simdi.year
    yas = bu_yil - dogum_yili
    print(f"{yas} Yaşındasınız.")
    if yas < 18:
        print("Reşit Değilsiniz.")
    elif yas>=18 and yas<=65:
        print("Yetişkinsiniz")
    else:
        print("Emeklilik yaşındasınız.")
except ValueError as e:
    print(f"Hata: {e}")
finally:
    print("İşlem tamamlandı.")
```
