```python
import random 
secenekler = ['tas', 'kagit', 'makas']
print('*****Taş-Kağıt-Makas Oyununa Hoşgeldiniz.*****')
while True:
    try:
        bilgisayar_secim = random.choice(secenekler)
        kullanici_secim = input('Tas, Kagit veya Makas seçiminiz çıkmak için q: ').lower()
        if kullanici_secim == 'q':
            print('Oyun bitmiştir, kendinize iyi bakın.!!!')
            break 
        elif kullanici_secim not in secenekler:
            raise ValueError ('Lütfen tas, kagit veya makas"tan birini seçiniz.')   
        else:
            if kullanici_secim == bilgisayar_secim:
                print('Beraberlik')
            elif (kullanici_secim == 'tas' and bilgisayar_secim == 'makas') or (kullanici_secim == 'makas' and bilgisayar_secim == 'kagit') or (kullanici_secim == 'kagit' and bilgisayar_secim == 'tas'):
                print(f'Kullanıcı Seçim: {kullanici_secim}, Bilgisayar Seçim: {bilgisayar_secim}')
                print('Kullanıcı Kazandı.')
            else:
                print(f'Kullanıcı Seçim: {kullanici_secim}, Bilgisayar Seçim: {bilgisayar_secim}')
                print('Bilgisayar Kazandı')
    except ValueError as e:
        print(f'Hata: {e}')
```
