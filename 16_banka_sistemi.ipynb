```python
class Hesap():
    def __init__(self, sahip, bakiye = 0): # başlangıçta herkesin bakiyesi = 0
        self.sahip = sahip
        self.bakiye = bakiye
    def para_yatir(self, miktar):
        if miktar < 0:
            print('Negatif tutarda para yatırma işlemi yapamazsınız.')
        elif miktar == 0:
            print('Sıfır tutarında para yatırma işlemi yapamazsınız.')
        else:
            self.bakiye += miktar
            print(f'{miktar} TL hesabınıza yatırıldı. Güncel bakiyeniz: {self.bakiye} TL')
    def para_cek(self, miktar):
        if miktar < 0:
            print('Negatif TL tutarında para çekemezsiniz.')
        elif miktar == 0:
            print('Sıfır TL tutarında para çekemezsiniz.')
        else:
            if self.bakiye < miktar:
                print(f'İstediğiniz tutarda para çekmek için yeterli bakiyeye sahip değilsiniz. Güncel Bakiye: {self.bakiye} TL')
            else:
                self.bakiye = self.bakiye - miktar
                print(f'Hesabınızdan {miktar} TL çekildi. Güncel Bakiye: {self.bakiye} TL. İyi günlerde harcayın ...')
class VadeliHesap(Hesap): # VadeliHesap, Hesap sınıfından türemektedir.
    def __init__(self, sahip, faiz_orani, bakiye = 0):
        super().__init__(sahip, bakiye)
        self.faiz_orani = faiz_orani
    def faiz_uygula(self):
        self.bakiye = (self.bakiye) + (self.bakiye * self.faiz_orani)
        print(f'% {self.faiz_orani * 100} faiz oranı ile hesabınıza faiz uygulandı.\n Güncel bakiyeniz: {self.bakiye} TL')
vadeli1 = VadeliHesap("Talha", 0.05, 1000)
vadeli1.faiz_uygula()
```
