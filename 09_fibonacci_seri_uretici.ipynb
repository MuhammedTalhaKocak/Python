```python
def fibo_seri(eleman_sayisi: int) -> list:
    if not isinstance(eleman_sayisi, int) or eleman_sayisi <= 0:
        raise ValueError('Geçersiz Eleman Sayısı Girdiniz.')   
    fibo_serisi = [0, 1]
    for _ in range(2, eleman_sayisi):
        fibo_serisi.append(fibo_seri_son_iki := fibo_serisi[-2] + fibo_serisi[-1])   
    return fibo_serisi[:eleman_sayisi]
```

```python
try:
    eleman_sayisi = int(input('Listenizin uzunluğunu giriniz: '))
    fibonacci_serisi = fibo_seri(eleman_sayisi)
    print(f'{eleman_sayisi} elemanlı fibonacci seriniz: {fibonacci_serisi}')
except ValueError as e:
    print(f'Hata: {e}')
finally:
    print('İşlem Tamamlandı.')
```
