```python
# Amaç: kullanıcının gelir/gider kayıtlarını tutan bir program yazmak.
import re
def ekle():
    try:
        secim = int(input('Gelir eklemek için [1], Gider eklemek için [2]"ye basınız.'))
        if secim not in [1,2]:
            raise ValueError ("Lütfen 1 veya 2'ye basınız.")
        elif secim == 1:
            gelir = int(input("Geliriniz: "))
            kaynak = input("Gelir Kaynağınız Nedir? ")
            with open("butce.txt", "a", encoding="utf-8") as dosya:
                dosya.write(f'Geliriniz: {gelir}, Kaynak: {kaynak}\n')
            print("Gelir Kaynağınız Eklendi.")
        elif secim == 2:
            gider = int(input("Gideriniz: "))
            kaynak = input("Gider Kaynağınız Nedir? ")
            with open("butce.txt", "a", encoding="utf-8") as dosya:
                dosya.write(f'Gideriniz: {gider}, Kaynak: {kaynak}\n')
            print("Gider Kaynağınız Eklendi.")
    except ValueError as hata:
        print(f'Hata: {hata}')
def rapor_gor():
    try:
        with open("butce.txt", "r", encoding="utf-8") as dosya:
            if len(satirlar := dosya.readlines()) == 0:
                raise ValueError('Dosya Tamamen Boş')
            else:
                gelirler = []
                giderler = []
                hedef_gelir_kelimesi = "Geliriniz"
                gelir_regex = rf"^{hedef_gelir_kelimesi}.*?(\d+)"
                hedef_gider_kelimesi = "Gideriniz"
                gider_regex = rf"^{hedef_gider_kelimesi}.*?(\d+)"
                for satir in satirlar:
                    if match := re.search(gelir_regex, satir):
                        gelirler.append(int(match.group(1)))
                    elif match := re.search(gider_regex, satir):
                        giderler.append(int(match.group(1)))
                toplam_gelir = sum(gelirler)
                toplam_gider = sum(giderler)
                net_bakiye = toplam_gelir - toplam_gider
                print(f"Toplam Gelir: {toplam_gelir}")
                print(f"Toplam Gider: {toplam_gider}")
                print(f"Net Bakiye: {net_bakiye}")
    except ValueError as e:
        print(f"Hata: {e}")
    except FileNotFoundError:
        print("Hata: Aramaya çalıştığınız dosya dizinde bulunamadı!")
def ana_program():
    print('Kişisel Bütçe Takipçisine Hoşgeldiniz.')
    while True:
        try:
            secim = input("""Yapmak İstediğiniz İşlemi Seçiniz
            Kayıt Eklemek için [1],
            Rapor Görmek için [2],
            Çıkmak için [3]'e basınız.""")
            if secim not in ["1","2","3"]:
                raise ValueError ('Geçersiz İşlem Seçimi !!!')
            elif secim == "3":
                print('Görüşmek üzere')
                break
            elif secim == "1":
                ekle()
            elif secim == "2":
                rapor_gor()
        except ValueError as e:
            print(f"Hata: {e}")
```

```python
ana_program()
```
