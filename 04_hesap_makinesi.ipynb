```python
def topla(a,b):
    sonuc = a+b
    return sonuc
def cikar(a,b):
    sonuc = a-b
    return sonuc
def bol(a,b):
    if b == 0:
        raise ValueError("Bölünen sayı 0 olamaz.")
    else:
        sonuc = a/b
        return sonuc
def carp(a,b):
    sonuc = a*b
    return sonuc

print("**********Hesap Makinesine Hoşgeldiniz**************")
while True:
    islem = input("""
    Lütfen Yapmak İstediğiniz İşlemi Seçiniz
    '+', '-', '*', '/'
    """)
    try:
            sayi1 = int(input("Lütfen 1.sayıyı giriniz: "))
            sayi2 = int(input("Lütfen 2.sayıyı giriniz: "))

            if islem not in ["+", "-", "*", "/"]:
                raise ValueError("Geçersiz İşlem Tipi")
            elif islem == "+":
                sonuc = topla(sayi1,sayi2)
            elif islem == "-":
                sonuc = cikar(sayi1,sayi2)    
            elif islem == "*":
                sonuc = carp(sayi1,sayi2)
            elif islem == "/":
                sonuc = bol(sayi1,sayi2)
            if sonuc is not None:
                print(f"Sonuç: {sonuc}")
    except ValueError as hata:
            print(f"Hata: {hata}")
    durum = input("Devam etmek isterseniz evet, devam etmek istemezseniz hayır yazın.")
    if durum.lower() == 'hayır':
         print("İşlem Sona Erdi.Hoşçakalın !!!") 
         break
    elif durum.lower() == 'evet':
         continue
        
```
