```python
import os 
try:
    klasor_yolu = "C:/Users/asus/Desktop/Dosyalar"
    dosyalar = os.listdir(klasor_yolu)
    print(dosyalar)
except FileNotFoundError as e:
    print(f"Hata: {e}")
```

```python
import shutil
for dosya in dosyalar:
    isim, uzanti = os.path.splitext(dosya)
    print(f'Dosya ismi: {isim}, Uzantısı: {uzanti}')
    klasor_adi = uzanti[1:].upper()
    hedef_klasor = os.path.join(klasor_yolu, klasor_adi)
    os.makedirs(hedef_klasor ,exist_ok=True)
    suan = os.path.join(klasor_yolu, dosya)
    hedef = os.path.join(klasor_yolu, klasor_adi, dosya)
    shutil.move(suan, hedef)
```
